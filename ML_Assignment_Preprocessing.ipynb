# Load the dataset

```python
import pandas as pd
from scipy.io import arff
```

```python
data, meta = arff.loadarff('Rice_Cammeo_Osmancik.arff')
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

# Data Quality Summary
Data quality was assessed during EDA. No missing values or duplicate records were identified, and the numerical features contained valid values. The target classes showed an approximately 57:43 distribution, which was not considered severe enough to require resampling initially. Statistical outliers were identified in several features and will be investigated further before determining an appropriate treatment.

# Train/Test split

```python
X = df.drop(columns=["Class"])
y = df["Class"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

## Verify the split

```python
print("Training set:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True) * 100)

print("\nTest set:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True) * 100)
```

# Outlier Strategy 
Statistical outliers were identified in several morphological features during EDA. Since these observations represent measured rice-grain characteristics and no evidence currently indicates that they are erroneous or invalid, the observations will be retained rather than removed. Their potential influence on model performance will be considered during model development.

# Correlation / Redundancy
EDA identified strong correlations among several numerical
features, particularly between Area and Convex_Area, and between
Perimeter and Major_Axis_Length.

Since highly correlated features may contain redundant information,
feature redundancy will be considered during feature selection.

Rather than removing correlated features solely based on the
correlation matrix, both the original feature set and a reduced
feature set will be evaluated during model development to determine
whether removing redundant features affects model performance.

# Feature Selection Plan

The dataset contains seven numerical input features. EDA identified
strong correlations between several features, particularly Area and
Convex_Area, and between Perimeter and Major_Axis_Length.

Feature selection will therefore be investigated using both
correlation-based analysis and a model-based selection approach.

First, a model using all seven features will be established as the
reference feature set. A reduced feature set will then be created
by removing features that show strong redundancy.

Forward Feature Selection will also be evaluated to identify a
subset of features that provides useful predictive performance.

The different feature sets will be compared during model
development using consistent validation and evaluation metrics.

# Scaling Strategy

The dataset contains numerical features with different scales, as well as statistical outliers and skewed distributions identified during EDA. Therefore, scaling will be applied to scale-sensitive models such as Logistic Regression, kNN, and SVM. StandardScaler will be considered as the initial scaling method, while RobustScaler will also be evaluated because the dataset contains outliers. The effect of the scaling strategy will be assessed during model development.

# Decision log
## Target Variable

The target variable is Class, which represents the rice variety:
Cammeo or Osmancik.

The remaining seven morphological measurements are used as input
features.

## Train/Test Split

The dataset will be divided into training and testing sets using
an 80:20 split.

Stratified splitting will be used to preserve the class distribution
of Cammeo and Osmancik in both sets.

The test set will remain untouched until final model evaluation.

## Missing Values

No missing values were identified during EDA.

Therefore, no missing-value imputation will be applied.

## Duplicate Records

No duplicate records were identified during EDA.

Therefore, no duplicate records will be removed.

## Outlier Strategy

Statistical outliers were identified in several morphological
features during EDA.

The observations will initially be retained because an IQR-based
outlier does not necessarily indicate an erroneous observation.
No evidence of invalid measurements was identified.

Their potential influence will be considered during model
development.

## Class Distribution

The target classes are not perfectly balanced, with approximately
57% Osmancik and 43% Cammeo.

Stratified splitting will therefore be used to preserve the class
distribution. No resampling will be applied initially.

## Feature Redundancy

EDA identified strong correlations between several features,
particularly Area and Convex_Area, and Perimeter and
Major_Axis_Length.

These relationships may indicate redundant information.

Rather than automatically removing correlated features, feature
selection will be evaluated during model development.

## Feature Selection

The original seven-feature dataset will be retained as the reference
feature set.

A reduced feature set will also be evaluated based on feature
redundancy identified during EDA.

Forward Feature Selection will be considered as a model-based
feature-selection approach, and the resulting feature subset will
be compared with the original feature set during model development.

## Scaling Strategy

The input features have different numerical scales. In addition,
EDA identified statistical outliers and skewed distributions.

Scaling will therefore be considered for scale-sensitive models
such as Logistic Regression, kNN and SVM.

StandardScaler will be evaluated as an initial scaling method.
Because statistical outliers are present, RobustScaler will also
be considered.

The effect of the scaling strategy will be evaluated during model
development rather than assuming that one scaling method is
universally optimal.

## Data Leakage Prevention

Any preprocessing operation that learns parameters from the data,
such as scaling or model-based feature selection, will be fitted
using only the training data or the appropriate training folds
during cross-validation.

The test set will not be used to fit preprocessing steps or select
features.
